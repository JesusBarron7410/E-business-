```
!pip install pycryptodome -q

from Crypto.Cipher import AES, PKCS1_OAEP
from Crypto.PublicKey import RSA
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
import base64

key = RSA.generate(2048)
private_key = key.export_key()
public_key = key.public_key().export_key()

print("1. Llaves RSA generadas correctamente.\n")

mensaje_original = "Este es un mensaje secreto para la actividad de Criptografía"
print(f"Mensaje Original: {mensaje_original}")

session_key = get_random_bytes(16)

cipher_aes = AES.new(session_key, AES.MODE_CBC)
iv = cipher_aes.iv
mensaje_cifrado_bytes = cipher_aes.encrypt(pad(mensaje_original.encode(), AES.block_size))

recipient_key = RSA.import_key(public_key)
cipher_rsa = PKCS1_OAEP.new(recipient_key)
enc_session_key = cipher_rsa.encrypt(session_key)

print(f"\n2. Mensaje cifrado con AES (Base64): {base64.b64encode(mensaje_cifrado_bytes).decode()}")
print(f"3. Llave AES cifrada con RSA: {base64.b64encode(enc_session_key).decode()[:50]}...")

code_private_key = RSA.import_key(private_key)
cipher_rsa_dec = PKCS1_OAEP.new(code_private_key)
decoded_session_key = cipher_rsa_dec.decrypt(enc_session_key)

cipher_aes_dec = AES.new(decoded_session_key, AES.MODE_CBC, iv)
mensaje_final = unpad(cipher_aes_dec.decrypt(mensaje_cifrado_bytes), AES.block_size).decode()

print(f"\n4. Proceso de descifrado completado.")
print(f"Resultado final: {mensaje_final}")
```
